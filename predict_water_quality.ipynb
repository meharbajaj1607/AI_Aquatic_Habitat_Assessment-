```python
import os
import joblib
import pandas as pd


# ============================================================
# MODEL PATHS
# ============================================================

MODEL_DIR = r"C:\Users\mehar\Desktop\AI_Aquatic_Habitat_Assessment (2)\AI_Aquatic_Habitat_Assessment\ml\models"

PH_MODEL_PATH = os.path.join(
    MODEL_DIR,
    "final_ph_gradient_boosting.pkl"
)

DO_MODEL_PATH = os.path.join(
    MODEL_DIR,
    "final_do_gradient_boosting.pkl"
)


# ============================================================
# LOAD MODELS
# ============================================================

ph_model = joblib.load(PH_MODEL_PATH)
do_model = joblib.load(DO_MODEL_PATH)

print("Models loaded successfully!")
```

```python
# ============================================================
# PREDICTION FUNCTION
# ============================================================

def predict_water_quality(
    water_temperature,
    sp_conductance,
    turbidity
):

    input_data = pd.DataFrame([{
        "water_temperature": float(water_temperature),
        "sp_conductance": float(sp_conductance),
        "turbidity": float(turbidity)
    }])

    predicted_ph = ph_model.predict(input_data)[0]
    predicted_do = do_model.predict(input_data)[0]

    return {
        "predicted_ph": round(float(predicted_ph), 3),
        "predicted_do": round(float(predicted_do), 3)
    }
```

```python
result = predict_water_quality(
    water_temperature=10.0,
    sp_conductance=0.5,
    turbidity=5.0
)

print(result)
```
